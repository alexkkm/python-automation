A hello world example of using python script to do automation

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# Create a new instance of the Chrome driver
driver = webdriver.Chrome()

# Navigate to a website
driver.get("https://www.example.com")

# Wait for the element to be present
try:
    element = WebDriverWait(driver, 10).until(EC.presence_of_element_located((By.CSS_SELECTOR, "h1")))
    print("Text on the element:", element.text)
except Exception as e:
    print("Could not find the element:", e)

# Close the browser
driver.quit()
```

An example to do web scapping from stock firm website

```python
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
import time

# Step 0 define the url
urlLink="https://www.laohu8.com/fund/HK0000369188"

# Step 1: Launch browser (Chrome in this case)
driver = webdriver.Chrome()

# Step 2: Open a website
driver.get(urlLink)

# Step 3: Print page content (Hello World style)
print("Page title is:", driver.title)

# Step 4: Find and print the text on the element with class="fund-nav increase"
try:
    element = driver.find_element(By.CSS_SELECTOR, ".fund-nav.increase")  # Example: div with class="fund-nav increase"
    navValueText = element.text
    print("Text on the element:", navValueText)
except Exception as e:
    print("Could not find the element:", e)

# Step 5: Wait a bit to see the result
time.sleep(3)

# Step 6: Close browser
driver.quit()
```
