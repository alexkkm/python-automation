### The basic script of calling the AI agent
```python
import os
import sys
from dotenv import load_dotenv
from openai import OpenAI

# Load environment variables from .env file
load_dotenv()

# Check API key BEFORE initializing client
api_key = os.getenv("openrouter_api_key")   # please ensure you create the .env file and add your OpenRouter API key as openrouter_api_key="YOUR_API_KEY"

# Check if API key is found
if not api_key:
    print("Error: API key not found in environment variables.")
    sys.exit(1)

# Initialize OpenRouter client with verified key
client = OpenAI(
    api_key=api_key,
    base_url="https://openrouter.ai/api/v1",
)

# Function to call OpenRouter API
def call_openrouter_api():

    # Call OpenRouter API 
    response = client.chat.completions.create(
        model="openrouter/free",
        messages=[
            {"role": "system", "content": "Please return Hello World."},
            {"role": "user", "content": "Hello World."}
        ],
        temperature=0.3,
    )

    # Print the response 
    print(response.choices[0].message.content)

# Main function
if __name__ == "__main__":
    call_openrouter_api()
```

````python
import os
import sys
from dotenv import load_dotenv
from openai import OpenAI

# Load environment variables from .env file
load_dotenv()

# Check API key BEFORE initializing client
api_key = os.getenv("openrouter_api_key")   # please ensure you create the .env file and add your OpenRouter API key as openrouter_api_key="YOUR_API_KEY"

# Check if API key is found
if not api_key:
    print("Error: API key not found in environment variables.")
    sys.exit(1)

# Initialize OpenRouter client with verified key
client = OpenAI(
    api_key=api_key,
    base_url="https://openrouter.ai/api/v1",
)

# Function to call OpenRouter API
def call_openrouter_api():

    # Call OpenRouter API 
    response = client.chat.completions.create(
        model="openrouter/free",
        messages=[
            {"role": "system", "content": "Please return Hello World."},
            {"role": "user", "content": "Hello World."}
        ],
        temperature=0.3,
    )
    
    # Print the response
    print(response.choices[0].message.content)

# Main function
if __name__ == "__main__":
    call_openrouter_api()
````
