```python
from openpyxl import Workbook
from openpyxl.utils import get_column_letter

# Create a new workbook
wb = Workbook()
ws = wb.active

# Add/Replace a value to cell A1
ws['A1'] = 'Testing'

# Read the value of cell A1
value = ws['A1'].value
print("Value of cell A1:", value)

# Save the workbook
wb.save('Example.xlsx')

# Close the workbook
wb.close()
```
